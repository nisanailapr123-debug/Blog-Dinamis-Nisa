Langkah 1 — Mengenali fitur & label.

```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label
```

Langkah 2 — Klasifikasi vs regresi (amati tipe label).

```python
# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

Langkah 3 — Pembagian data latih & uji.

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

3. Untuk kasus 'memprediksi jumlah pengunjung toko besok', tentukan jenis pembelajaran, subkategori, fitur, dan label.


```python
import pandas as pd

data = {
    'pengunjung_kemarin': [100, 120, 150, 130, 180],
    'promosi': ['Tidak', 'Ya', 'Ya', 'Tidak', 'Ya'],
    'hari_libur': ['Tidak', 'Tidak', 'Ya', 'Tidak', 'Ya'],
    'pengunjung_besok': [110, 140, 180, 125, 200]
}

df = pd.DataFrame(data)

print(df)

# Menentukan fitur dan label
X = df[['pengunjung_kemarin', 'promosi', 'hari_libur']]
y = df['pengunjung_besok']

print("\nFitur:")
print(X)

print("\nLabel:")
print(y)

print("\nJenis pembelajaran: Supervised Learning")
print("Subkategori: Regresi")
print("Alasan: label berupa jumlah pengunjung dalam bentuk angka.")
```
